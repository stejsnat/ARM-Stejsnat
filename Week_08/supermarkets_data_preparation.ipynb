# Importing and preparing supermarkets data

**This Jupyter notebook does the following:**

- Imports the supermarket data (JSON) from the Overpass Web API
- Converts the nested column `tags` (dictionaries) to separate columns and merges them with the main data
- Counts missing and duplicated values and checks the data types
- Saves the prepared data to `Data/supermarkets_data_prepared.csv`

## Libraries and settings

```python
# Libraries
import os
import fnmatch
import pandas as pd

# Ignore warnings
import warnings
warnings.filterwarnings("ignore")

# Get current working directory
print('Current working directory:', os.getcwd())

# Show .json files in the folder 'Data'
flist = fnmatch.filter(os.listdir('./Data'), '*.json')
for i in flist:
    print(i)
```

## Importing data

```python
# Read the data to a pandas data frame
df1 = pd.read_json('./Data/supermarkets.json', encoding='utf-8')
df1.head(5)
```

## Count number of rows and columns in the data frame

```python
# Dimension (rows, columns)
print('Dimension:', df1.shape)

# Number of rows
print('Number of rows:', df1.shape[0])

# Number of columns
print('Number of columns:', df1.shape[1])
```

## Column 'tags' is a pandas Series with dictionaries -> change to data frame

```python
# Type of the first item of column tags
print(type(df1.tags))
print(type(df1.tags[0]))

# Content of the first item of column tags
print(df1.tags[0].keys())

# Change to data frame
df2 = pd.DataFrame.from_records(df1.tags)
df2 = df2[['brand', 'shop', 'addr:city', 'addr:street', 'addr:housenumber', 'addr:postcode']]

# Rename selected columns
df2 = df2.rename(columns={'addr:city': 'city',
                          'addr:street':'street',
                          'addr:housenumber': 'housenumber',
                          'addr:postcode': 'postcode'})

# Show first records of data frame
df2.head()
```

## Merge df1 and df2

```python
# Merge df and df2
df = pd.merge(df1[['type', 'id', 'lat', 'lon']], 
              df2[['brand', 'shop', 'city', 'street', 'housenumber', 'postcode']],
              left_index=True, 
              right_index=True)
df.head(5)
```

## Count and identify the number of missing values (if any)

```python
# Count missing values
print(pd.isna(df).sum())

# Identify rows with missing values, e.g.:
df.loc[pd.isna(df['city'])]
```

## Count and identify duplicated values (if any)

```python
# Count duplicated values
print(df.duplicated().sum())

# Identify rows with duplicated values, e.g.:
df[df[['id']].duplicated()]
```

## Get data types of all variables

```python
# Get data types (note that since pandas 3.0, text columns have the data type 'str';
# in older versions of pandas, they have the data type 'object')
df.dtypes
```

### Save data to file

```python
df.to_csv('./Data/supermarkets_data_prepared.csv', 
          sep=",", 
          encoding='utf-8',
          index=False)
```

### Include opening hours as additional variable in the data frame 

```python
df2 = pd.DataFrame.from_records(df1.tags)
df2 = df2[
    ['brand', 'shop', 'addr:city', 'addr:street',
     'addr:housenumber', 'addr:postcode', 'opening_hours']
]

df2 = df2.rename(columns={
    'addr:city': 'city',
    'addr:street': 'street',
    'addr:housenumber': 'housenumber',
    'addr:postcode': 'postcode'
})

df = pd.merge(
    df1[['type', 'id', 'lat', 'lon']],
    df2[['brand', 'shop', 'city', 'street', 'housenumber', 'postcode',
         'opening_hours']],
    left_index=True,
    right_index=True
)
```

```python
display(df[['brand', 'city', 'street', 'housenumber', 'postcode', 'opening_hours']].head(10))
```

### Filter supermarkets with available opening hours.

```python
# Filter supermarkets that have opening hours
df_with_opening_hours = df.loc[
    df['opening_hours'].notna() &
    df['opening_hours'].astype(str).str.strip().ne('')
]

# Show the number of matches and the first 10 results
print(f"Supermarkets with available opening hours: {len(df_with_opening_hours)}")
display(df_with_opening_hours.head(10))
```
