## Use Case 1: Price prediction model for rental apartments in the canton of Zurich

**1. Problem definition**
- Rents in the canton of Zurich are hard to compare. Landlords and tenants often estimate prices based on gut feeling.
- As a result, apartments are listed too high or too low. This leads to longer vacancies or lost income, and tenants cannot judge whether a price is fair.

**2. Objective**
- Develop a prediction model that estimates the market rent of an apartment as accurately as possible, based on its features (location, size, number of rooms, year of construction, equipment).

**3. Research question**
- Which features influence rental prices in the canton of Zurich the most, and how accurately can the rent be predicted with them?

**4. Hypothesis**
- H1: Living area and location (municipality or distance to the city of Zurich) explain most of the differences in rental prices.
- H2: A model with several variables (e.g. multiple regression) predicts rents clearly better than a model that only uses living area.

**5. Expected added value**
- Landlords and property managers: realistic pricing, fewer vacancies, higher returns.
- Tenants: more transparency about whether an offer is fair.
- Real estate platforms and investors: automated valuation and better market analysis.

# Use Case 2: Demand and supply for supermarkets in Swiss municipalities

**1. Problem definition**
- Some municipalities have too few supermarkets, so people must travel far to shop. Others have too many, which creates strong competition.
- Retailers often do not know exactly where a new store would be profitable.

**2. Objective**
- Compare demand (population, purchasing power) with supply (number and sales area of supermarkets) in each municipality, and identify municipalities that are over- or undersupplied.

**3. Research question**
- How are population size, income and settlement type related to supermarket density, and in which municipalities does supply differ strongly from expected demand?

**4. Hypothesis**
- H1: The larger the population and the higher the purchasing power of a municipality, the more supermarket space per person it has.
- H2: Rural municipalities are more likely to be undersupplied, while urban municipalities are more likely to be oversupplied.

**5. Expected added value**
- Retailers: data-based location decisions and lower investment risk.
- Municipalities and planners: identify supply gaps and support spatial planning.
- Society: better basic supply and shorter shopping distances.

## Use Case 1: Rental Price Prediction, Canton of Zurich

**Main source: Flatfox.ch (listing-level data per offer)**
- **Target variable:** Net rent (alternatively gross rent). Record the additional costs so prices are comparable.
- **Property features:** Living area (m²), number of rooms, floor, year of construction, property type (apartment, penthouse, etc.).
- **Amenities:** Balcony, elevator, washing machine, parking space, pets allowed.
- **Location:** Address, postal code, town and coordinates (if available).
- **Other:** Availability date and listing date.

*Rationale:* Area, number of rooms and location explain rent the most, so the model needs these features as inputs.

**Supplement: Regionalporträts 2021 (SFSO), municipality level**
- Population and population density.
- Age structure and share of foreign residents.
- Household size.
- Taxable income per taxpayer.
- Vacancy rate and housing stock or new construction activity.
- Jobs (by sector).

*Rationale:* These variables describe demand and supply on the housing market. A municipality with high income and a low vacancy rate usually has higher rents. A listing alone says nothing about this.

**Linking:** Via municipality name or SFSO municipality number (Flatfox postal code/town → municipality). Then filter the dataset to the canton of Zurich.

## Use Case 2: Demand and Supply of Supermarkets in Swiss Municipalities

**Main source: OpenStreetMap (`shop=supermarket`, queried via Overpass API or osmnx)**
- Coordinates (lat/lon).
- Name, brand and operator (Migros, Coop, Denner, Aldi, Lidl, etc.).
- Address (`addr:*`) and opening hours (`opening_hours`).
- Optionally `shop=convenience` to distinguish small shops.
- Derived: number of supermarkets per municipality and per 1,000 inhabitants (**supply**).

**Supplement: Regionalporträts 2021 (SFSO), municipality level**
- Population, density and households (**demand**).
- Taxable income as a proxy for purchasing power.
- Age structure and share of foreign residents (consumption behavior).
- Settlement area, jobs and commuters (daytime population).

*Rationale:* Demand depends on residents and purchasing power. Comparing it with supply shows over- and undersupplied municipalities.

**Linking:** Assign the OSM points to municipality boundaries via spatial join (e.g. swisstopo swissBOUNDARIES3D) and aggregate per municipality. Then merge with the Regionalporträts via the SFSO municipality number.

## Points for Discussion
- OSM can be incomplete, and it has no data on sales area or revenue.
- The Regionalporträts date from 2021. Flatfox and OSM data are current, so there is a time lag.
- Flatfox shows asking rents (new lettings), not existing-tenancy rents.
- Check the variables in the `data_for_import` sheet before merging. Not all of the listed indicators are guaranteed to be included there.
